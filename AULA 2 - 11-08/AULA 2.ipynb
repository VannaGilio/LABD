### Correção da Atividade Diagnóstica

Subconjuntos do SQL:

- DDL (Data Definition Language):
Se refere a um conjunto de comandos SQL usados para definir e gerenciar a estrutura de bancos de dados. 

Os principais comandos DDL: 
- CREATE(Criar tabelas...) 
- TRUNCATE(Apagar todos os dados de uma tabela e mantem a estrutura)
- ALTER(Modifica a estrutura de objetos existentes)
- DROP(Excluir)

---
DML (Data Manipulation Language):
Conjunto que é utilizado para gerenciar os dados do banco de dados.

Principais comandos DML:
- SELECT
- INSERT
- UPDATE
- DELETE

---
DCL (Data Control Language):
- Função de controlar permissões e segurança no banco de dados.

Principais comandos DCL:
- GRANT(Concede permissões a usuários/roles)
- REVOKE(Retira permissões)

---
TCL (Transaction Control Language):
Função é gerenciar transações e garantir integridade dos dados.

Principais comandos TCL:
- COMMIT (Confirma a transação)
- ROLLBACK (Desfaz alterações não confirmadas)
- SAVEPOINT (Cria pontos de restauração dentro de uma transação)
